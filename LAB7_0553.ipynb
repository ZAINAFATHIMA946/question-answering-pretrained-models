**LAB 7: Question Answering using Pretrained Models: Use a SQuAD-style context-question-answer pipeline with BERT. Tools: Hugging Face, pipeline(task="question-answering").**

Objective

To implement a Question Answering system using a pretrained BERT model via the Hugging Face pipeline and extract answers from a given context.

Question Answering (QA) is an NLP task where a model answers a question based on a given passage (context).

🔹 Types:
Extractive QA → Answer is a span from the context (used here)

Generative QA → Answer is generated


🤖 Model Used
BERT (Bidirectional Encoder Representations from Transformers)
Pretrained on SQuAD dataset (Stanford Question Answering Dataset)


```python
# Step 1: Install Libraries (Force v4 to support "question-answering" pipeline)
!pip install "transformers<5.0.0"

# Step 2: Import Pipeline
from transformers import pipeline

# Step 3: Load Pretrained QA Model
qa_pipeline = pipeline("question-answering")

# Step 4: Define Context (SQuAD style)
context = """
Artificial Intelligence (AI) is the simulation of human intelligence processes by machines,
especially computer systems. These processes include learning, reasoning, and self-correction.
AI is widely used in applications such as natural language processing, robotics, and computer vision.
"""

# Step 5: Define Question
question = "What are the processes included in AI?"

# Step 6: Get Answer
result = qa_pipeline(question=question, context=context)

# Step 7: Display Output
print("Question:", question)
print("Answer:", result['answer'])
print("Confidence Score:", result['score'])
```

**ACTIVITY** 5**: Multiple Questions on Same Context

🎯 Aim:
To evaluate model performance on multiple queries.

❓ Task:
Given a context on Machine Learning, ask at least 5 questions.

💡 Hint:
Use a loop to process multiple questions.

```python
# ACTIVITY 5: Multiple Questions on the Same Context

context = """
Machine Learning (ML) is a subset of Artificial Intelligence that
enables computers to learn patterns from data and make predictions
without being explicitly programmed for every task.

The three common types of Machine Learning are supervised learning,
unsupervised learning, and reinforcement learning.

Supervised learning uses labeled data to train a model.
Unsupervised learning discovers patterns in unlabeled data.
Reinforcement learning trains an agent through rewards and penalties.

Machine Learning is used in image recognition, spam detection,
recommendation systems, and fraud detection.
"""

# Define five questions
questions = [
    "What is Machine Learning?",
    "What are the three types of Machine Learning?",
    "What does supervised learning use?",
    "What does unsupervised learning discover?",
    "Where is Machine Learning used?"
]

# Process multiple questions using a loop
for i, question in enumerate(questions, start=1):
    result = qa_pipeline(
        question=question,
        context=context
    )

    print(f"Question {i}: {question}")
    print("Answer:", result["answer"])
    print("Confidence Score:", round(result["score"], 4))
    print("-" * 60)
```

**ACTIVITY** 6: Domain-Specific QA

🎯 Aim:
To test QA on specific domains.

❓ Task:
Use context from:

Healthcare
Finance
Education

Ask domain-related questions.

```python
# ACTIVITY 6: Domain-Specific QA

# Define domain-specific contexts and questions
domains = {
    "Healthcare": {
        "context": """
        Artificial Intelligence is used in healthcare to assist doctors
        in analyzing medical images and identifying possible diseases.
        Machine Learning models can help predict health risks using
        patient data. AI tools can also support hospital administration.
        """,
        "questions": [
            "How is AI used in healthcare?",
            "What can Machine Learning models help predict?"
        ]
    },

    "Finance": {
        "context": """
        Artificial Intelligence is used in finance to detect fraudulent
        transactions and analyze financial data. Machine Learning models
        can help banks assess credit risk. Automated systems can also
        support customer service and financial forecasting.
        """,
        "questions": [
            "How is AI used in finance?",
            "What can Machine Learning models help banks assess?"
        ]
    },

    "Education": {
        "context": """
        Artificial Intelligence is used in education to support
        personalized learning and provide automated feedback.
        AI-based educational platforms can recommend learning materials
        based on student performance. Teachers can use these systems
        to identify topics where students need additional support.
        """,
        "questions": [
            "How is AI used in education?",
            "What can AI-based platforms recommend?"
        ]
    }
}

# Process each domain
for domain, data in domains.items():
    print("\n" + "=" * 60)
    print("DOMAIN:", domain)
    print("=" * 60)

    for question in data["questions"]:
        result = qa_pipeline(
            question=question,
            context=data["context"]
        )

        print("Question:", question)
        print("Answer:", result["answer"])
        print("Confidence Score:", round(result["score"], 4))
        print("-" * 50)
```

**ACTIVITY** 7: Case Sensitivity Check

🎯 Aim:
To analyze effect of uppercase/lowercase text.

❓ Task:
Change context case and observe results.*italicized text*

```python
# ACTIVITY 7: Case Sensitivity Check

# Original context
context_normal = """
Artificial Intelligence is used in healthcare to analyze medical images.
It is also used in education to support personalized learning.
"""

# Uppercase context
context_upper = context_normal.upper()

# Lowercase context
context_lower = context_normal.lower()

question = "Where is Artificial Intelligence used?"

# Store the different contexts
contexts = {
    "Normal Case": context_normal,
    "Uppercase": context_upper,
    "Lowercase": context_lower
}

# Test the model on each context
for name, context in contexts.items():
    result = qa_pipeline(
        question=question,
        context=context
    )

    print("\nContext Type:", name)
    print("Question:", question)
    print("Answer:", result["answer"])
    print("Confidence Score:", round(result["score"], 4))
    print("-" * 60)
```
